```python
import numpy as np
```

```python
# Q.1: Broadcasting:

# Create a NumPy array containing [10, 20, 30, 40, 50]. Multiply every element by 3 and then add 5.

numbers = np.array([10, 20, 30, 40, 50])

result = (numbers * 3) + 5

print("Result:",result)
```

```python
# Q.2: Array Splitting:

# Create a NumPy array containing [10, 20, 30, 40, 50, 60, 70, 80] and split it into 4 equal parts.

numbers = np.array([10, 20, 30, 40, 50, 60, 70, 80])

result = np.split(numbers,4)

print("Before Splitting:", numbers)
print("After Splitting:", result)
```

```python
# Q.3: Random Number Generation:

# Generate a 2D NumPy array with 3 rows and 4 columns containing random integers between 1 and 50 (inclusive). = inclusive means, 1 pani aauna sakxa ra 50 pani aauna sakxa.

arr = np.random.randint(1, 51, size=[3, 4]) # 51 because question ma inclusive xa so 50 pani aauna sakxa tei vayara 1 badi lene..

print("Random 2D array:")
print(arr)
```

```python
# Q.4: Statistical Functions:

# Create a NumPy array containing [12, 18, 24, 30, 36] and calculate its sum, mean, median, and standard deviation.

numbers = np.array([12, 18, 24, 30, 36])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Median:", np.median(numbers))
print("S.D:", np.std(numbers))
```

```python
# Q.5: Matrix Operation:

# Create two NumPy matrices, A = [[1, 2], [3, 4]] and B = [[5, 6], [7, 8]]. Perform matrix multiplication of A and B using NumPy.

mat_a = np.array([[1, 2], 
                  [3, 4]
                 ])
mat_b = np.array([[5, 6], 
                  [7, 8]
                 ])
result = mat_a @ mat_b

print("Mul of A and B:")
print(result)
```

```python
# Q.6: Array Joining:

# Create two NumPy arrays, a = [5, 10, 15] and b = [20, 25, 30]. Join them into a single array using np.concatenate().

arr_a = np.array([[5, 10, 15], [20, 25, 30]])

result = np.concatenate(arr_a)

print("Before Concatenate:") 
print(arr_a)
print("After Concatenate:", result)
```

```python
# Q.7: Broadcasting:

# Create a 2D NumPy array [[10, 20, 30], [40, 50, 60]]. Subtract [1, 2, 3] from each row using broadcasting.

# Method 1:
arr = np.array([[10, 20, 30],[40, 50, 60]])

result = arr - [1, 2, 3]

print("Method1 After Sub:")
print(result)

print()

# Method 2:
arr = np.array([[10, 20, 30],[40, 50, 60]])

sub_value = np.array([1, 2, 3])

print("Method2 After Sub:") 
print(arr-sub_value)
```

```python
# Q.8: Random Number Generation:

# Use np.random.choice() to randomly select 3 elements from the array [10, 20, 30, 40, 50, 60] without repeating any element.

# Method 1: Replace=False use na garne - means, same no. repeat huna sakxa.
numbers = np.array([10, 20, 30, 40, 50, 60])
rand_selection = np.random.choice(numbers, 3)
print("Random 3 Elements:", rand_selection)

print()

# Method 2: Replace=False use garne - means, same no. repeat hudaina.
numbers = np.array([10, 20, 30, 40, 50, 60])
rand_sel = np.random.choice(numbers, 3, replace=False)
print("Random 3 Elements:", rand_sel)
```

```python
# Q.9: Liner Algebra:

# Create a NumPy matrix [[4, 7], [2, 6]] and calculate its determinant using NumPy.

mat = np.array([[4, 7], [2, 6]])

result = np.linalg.det(mat)

print("Determinant:", result)
```

```python
# Q.10: Arithmetic Functions:

# Create a NumPy array [5, 10, 15, 20, 25] and calculate the cumulative sum using NumPy.
# 5, 15, 30, 50, 75 

numbers = np.array([5, 10, 15, 20, 25])

result = np.cumsum(numbers)

print("Cumulative Sum:", result)
```

```python
# Q.11: Array joining and splitting:

# Create a NumPy array containing [10, 20, 30, 40, 50, 60, 70, 80] and split it into 2 equal parts. Then print both parts separately.

numbers = np.array([10, 20, 30, 40, 50, 60, 70, 80])
part1, part2 = np.split(numbers, 2)
print("Part 1:", part1)
print("Part 2:", part2)
```

```python
# Q.12: Matrix Operation:

# Create a NumPy matrix [[2, 4, 6], [8, 10, 12]] and calculate its transpose using NumPy.

matrix = np.array([[2, 4, 6], [8, 10, 12]])
result = matrix.T
print("After Transpose:")
print(result)
```

```python
# Q.13: Arithmetic Function:

# Create a NumPy array [4, 9, 16, 25, 36] and calculate the square root of each element using NumPy.

numbers = np.array([4, 9, 16, 25, 36])
result = np.sqrt(numbers)

print("Numbers are:", numbers)
print("Square Root:",result)
```

```python
# Q.14: Broadcasting and Vectorization:

# Create a NumPy array [2, 4, 6, 8, 10]. Square every element and then subtract 1 from each result without using a loop.

numbers = np.array([2, 4, 6, 8, 10])
result = (numbers ** 2) - 1

print("Numbers:", numbers)
print("Square after sub 1:",result)
```

```python
# Q.16: Statistical Functions:

# Create a NumPy array [10, 20, 30, 40, 50, 60] and calculate its variance using NumPy.

numbers = np.array([10, 20, 30, 40, 50, 60])
result = np.var(numbers)

print("Numbers:", numbers)
print("Variance:", result)
```

```python
# Q.17: Linear Algebra:

# Create a NumPy matrix [[2, 1], [5, 3]] and calculate its inverse using NumPy.

mat = np.array([[2, 1], 
                [5, 3]
               ])
inverse = np.linalg.inv(mat)

print(inverse)
```
