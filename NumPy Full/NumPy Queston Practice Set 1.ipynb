```python
# Q.2.1: Broadcasting:

# Create a 2D NumPy array [[10, 20, 30], [40, 50, 60], [70, 80, 90]]. Using broadcasting, subtract [1, 2, 3] from every row of the 2D array.

arr = np.array([[10, 20, 30], 
                [40, 50, 60], 
                [70, 80, 90]
               ])
sub_value = np.array([1, 2, 3])

result =  arr - sub_value

print("Before Sub:")
print(arr)
print()
print("After Sub:")
print(result)
```

```python
# Q.3.1: Array Manipulation + Boolean Filtering + Statistical Functions

# Create a NumPy array containing the numbers from 1 to 12. Reshape it into a 3*4 matrix. Select all elements greater than 6, then calculate their sum and mean.

num = np.arange(1, 13)
reshapes = np.reshape(num, [3, 4])
print("After Reshapes:")
print(reshapes)

print()

num = np.arange(1, 13)
result = num[num>6]
c = np.sum(result)
res = np.mean(result)

print("Sum:",c)
print()
print("Mean:", res)
```

```python
# Q.3.2 : Random Number Generation + Sorting + Statistical Functions:

# A: Generate a 1D NumPy array containing 10 random integers between 10 and 100.
# B: Sort the array in descending order.
# C: Find the maximum, minimum, and mean of the original array.

rand_no = np.random.randint(10, 100, size=10)
print("Rand.No:",rand_no)
print("After Sorting:", np.sort(rand_no))
print("Max:", np.max(rand_no))
print("Min:", np.min(rand_no))
print("Mean:", np.mean(rand_no))
```

```python
# Q.3.3: Matrix Operations + Linear Algebra:

# Create two NumPy matrices A = [[2, 3], [4, 5]], B = [[1, 2], [3, 4]], 
# Perform the following operations:
#1. Matrix multiplication (A × B).
#2. Find the determinant of matrix A.
#3. Find the inverse of matrix A.

mat1 = np.array([[2, 3], [4, 5]])
mat2 = np.array([[1, 2], [3, 4]])

mat_mul1 = np.dot(mat1, mat2)
mat_mul2 = np.linalg.det(mat1)
mat_mul3 = np.linalg.inv(mat1)

print("Matrix Multiplication:")
print(mat_mul1)
print()
print("Determinant:",mat_mul2)
print()
print("Inverse:")
print(mat_mul3)
```

```python
# Q.3.3.4: Joining & Splitting + Array Manipulation:

# Create two NumPy arrays:a = [10, 20, 30], b = [40, 50, 60]. Join both arrays horizontally. Split the joined array into 3 equal parts. Display all three split arrays.

a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

joi = np.concatenate((a, b))
print("After Joining:", joi)

part1, part2, part3 = np.split(joi, 3)
print("Part1:", part1)
print("Part2:", part2)
print("Part3:", part3)
```

```python
# Q.3.5: Broadcasting + Vectorization + Statistical Functions:

# An online store has product prices and discount amount. Apply the discounts using NumPy broadcasting, calculate the final prices, and find the total and average final price.
# Given data are: prices = np.array([1000, 2000, 3000, 4000]) , discount = np.array([10, 20, 15, 25])

prices = np.array([1000, 2000, 3000, 4000])
discount = np.array([10, 20, 15, 25])
print("Original Prices:", prices)

final_prices = prices - discount
print("After Discount:",final_prices)

total = np.sum(final_prices)
print("Total of final prices:", total)

average = np.mean(final_prices)
print("Avg of final prices:", average)
```

```python
# Q.3.5: Broadcasting + Vectorization + Statistical Functions:

# An online store has product prices and discount percentage. Apply the discounts using NumPy broadcasting, calculate the final prices, and find the total and average final price.
# Given data are: prices = np.array([1000, 2000, 3000, 4000]) , discount = np.array([10, 20, 15, 25])

prices = np.array([1000, 2000, 3000, 4000])
discount = np.array([10, 20, 15, 25])
print("Original Prices:", prices)

discount_amt = discount/100 * prices
print("Discount Amt:", discount_amt)

final_prices = prices - discount_amt
print("After Discount:",final_prices)

total = np.sum(final_prices)
print("Total of final prices:", total)

average = np.mean(final_prices)
print("Avg of final prices:", average)
```
