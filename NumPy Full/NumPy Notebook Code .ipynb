```python
# Main operations in Data Science

salary = np.array([20000, 30000, 10000, 40000, 25000])

print("Total Salary:", np.sum(salary)) # sabai salary lai add garxa.
print("Average Salary:", np.mean(salary))
print("Maximum Salary:", np.max(salary))
print("Minimum Salary:", np.min(salary))
print("Standard Deviation:", np.std(salary))
print("Median:", np.median(salary))
print("Variance:", np.var(salary))
print("Sort:",np.sort(salary))  # Salary lai sano dekhi thulo kramma milaune
print("Unique:",np.unique(salary))  # Duplicate salary values hataera unique values dekhaune
print("Re-shape:",salary.reshape(5, 1))  # 5 rows ra 1 column bhayeko array banaune
print("Dimension:",np.ndim(salary))  # Array kati dimension ko chha bhanera herne
print("Array-Shape:",salary.shape)  # Array ma kati elements chhan bhanera herne
print("Array-Size:",salary.size)  # Array ma bhayeka total elements ganne
print("Index of Max value:",np.argmax(salary))  # Sabai bhanda thulo salary ko index nikalne
print("Index of Min value:",np.argmin(salary))  # Sabai bhanda sano salary ko index nikalne
print("25000 vanda badi ko index:",np.where(salary > 25000))  # 25000 bhanda badi salary ko index nikalne
print("25000 vand badi salary:",salary[np.where(salary > 25000)])  # 25000 bhanda badi salary values dekhaune
```

```python
# indexing ko main operations
```

```python
# Operation 1: Positive Indexing, array ko starting bata value access garne.

salary = np.array([20000, 30000, 10000, 40000, 25000])

print(salary[0])  # 1st employee ko salary nikalne.
print(salary[1])  # 2nd employee ko salary nikalne.
print(salary[2])  # 3rd employee ko salary nikalne.
print(salary[3])  # 4th employee ko salary nikalne.
print(salary[4])  # 5th employee ko salary nikalne.
```

```python
# Operation 2: Negative Indexing, array ko last bata value access garne.

salary = np.array([20000, 30000, 10000, 40000, 25000])

print(salary[-1])  # Last employee ko salary nikalne.
print(salary[-2])  # Last bata 2nd salary nikalne.
print(salary[-3])  # Last bata 3rd salary nikalne.
print(salary[-4])  # Last bata 4th salary nikalne.
print(salary[-5])  # Last bata 5th salary nikalne.
```

```python
# Operation 3: Update value, index use garera value update garne.

salary = np.array([20000, 30000, 10000, 40000, 25000])

salary[1] = 15000 # index 1 ma vako salary update vayara 15000 hunxa.
salary[3] = 35000
print(salary)
```

```python
# Operation 4: Index use garera value access garne.
             # kunai value lai variable ma rakhera pani use garna milxa.

salary = np.array([20000, 30000, 10000, 40000, 25000])

print(salary[2]) # without using variable index 2 ko value access garne.

employee_salary = salary[2] # Using variable, employee_salary variable ma value rakheko.
print("employee_salary:", employee_salary)
```

```python
# Operation 5: Arithmatic, indexing bata arithmatic garne.

salary = np.array([20000, 30000, 10000, 40000, 25000])

a = salary[0] + salary[1]  # 1st and 2nd salary add garne.
print("a:", a)

b = salary[2] + 40000      # index 2 ma vako salary ma 40000 add gareko.
print("b:", b)

c = salary[4] * 4          # index 4 ma vako salary lai 4 lay multiply gareko.
print("c:", c)

d = salary[3] / salary[2]  # index 3 ma vako salary lai index 2 ma vako salary lay divvide gareko.
print("d", d)
```

                           2.2 Slicing ?
=   Slicing vaneko NumPy Array bata chieyako vag or multiple/dherai      elements aakai patak nikalne process ho.

2.2.1, Why needed ?
- Dherai employee bata kehi employee ko salary nikalna.
- Data ko first, middle or last vag xanna.
- ML ma needed data ko subset banauna.
- Large data bata specific records herna.

example: Company ma 1000 employee xan vane first 10 jana or index 100 to index 200 samma employee ko data separate garna slicing use garienxa.

2.2.2, Syntax
      array[start:stop:step]  # start = kaha bata, stop = kaha samma (stop samabesh hudaina), step = kati gap ma
- Start = Kun index bata suru garne.
- Stop  = kun index aagadi rokine, stop ko index include hudaina.
- Step  = Kati kati index ko gap ma elements line.

```python
salary = np.array([20000, 30000, 10000, 40000, 25000])
print(salary)
```

```python
# Types of slicing
```

```python
# Type 1: Basic Slicing - Start and Stop, stop wala index include hudai na.

salary = np.array([20000, 30000, 10000, 40000, 25000])
print(salary)

print(salary[1:4]) # 1 to 3 index matra print garxa.

a = salary[1:3] # 1 to 2 index matra print garxa.
print("a",a)
```

```python
# Type 2: Start index na dine, 

salary = np.array([20000, 30000, 10000, 40000, 25000])
print(salary)

print(salary[:1]) # 0 index print garxa.
print(salary[:3]) # 0 index to index 2 samma dinxa.
print(salary[:4]) # 0 index to index 3 samma dinxa.
```

```python
# Type 3: Stop index na dine.

salary = np.array([20000, 30000, 10000, 40000, 25000])
print(salary)

print(salary[1:]) # index 1 to last index samma dinxa.
print(salary[3:]) # index 3 to last index samma dinxa.
```

```python
# Type 4: Start and Stop both na dine.

salary = np.array([20000, 30000, 10000, 40000, 25000])
print(salary)

print(salary[:])  # Pura array ko slice nikalne
```

```python
# Type 5: Step/Gap use garne

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

print(salary[::2]) # 1/1 index ko gap ma salary nikalxa.

print(salary[::3]) # 2/2 index ko gap ma salary nikalxa.

print(salary[::4]) # 3/3 index ko gap ma salary nikalxa.
```

```python
# Type 6: Start, Stop and Step all use garne.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

print(salary[2:8:2]) # index 2 bata start hunxa, index 8 samma janxa tara 7 samma matra nikalxa hunxa with 1/1 ko gap ma.

print(salary[0:9:3]) # index 0 bata start hunxa, index 9 samma janxa tara 8 samma matra nikalxa, ra 2/2 ko gap ma nikalxa.
```

```python
# Type 7: Negative index use garne.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

print(salary[-4:-1]) # last bata index -4 to index -2 samma ko value nikalxa.
print(salary[-8:-1]) # last bata index -8 to index -2 samma ko value nikalxa.
print(salary[-2:-8]) # wrong
```

```python
# Type 8: Negative step Scaling, reverse scaling.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

print(salary[::-1]) # last bata 0/0 index ko gap ma salary nikalxa.

print(salary[::-2]) # Last bata 1/1 index ko gap ma salary nikalxa.

print(salary[::-5]) # Last bata 4/4 index ko gap ma salary nikalxa.

print(salary[-5:]) # Last ko 5 ota index ko salary nikalxa.

print(salary[:-5]) # 1st index to -6 index ko salary nikalxa but -5 include hudaina.
```

```python
# Type 9: Negative step with start and stop.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

print(salary[8:1:-1]) # index 8 to index 2 samma ko salary reverse ma nikalxa.

print(salary[9:0:-2]) # index 9 to index 1 samma ko salary reverse ma nikalxa.
```

```python
# Type 10: Slicing bata new array banaune.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

selected_salary = salary[1:6] # index 1 to index 5 samma ko slice variable ma rakhxa.
print("Selected_Salary:", selected_salary)

last_3_salary = salary[-3:] # Last 3 index ko value lai slice variable ma rakhnxa.
print("Last_3_Salary:", last_3_salary)

reverse_salary = salary[::-1] # sabai index ko value lai reverse ma nikalxa
print("Reverse_Salary:", reverse_salary)
```

```python
# Type 11: Slicing bata dherai index ko value lai aaru same value lay change garne.

# Numpy ko basic slicing lay mainly original array kai view dinxa but slice bata value change garda original array pani change huna sakxa.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

salary[1:3] = 0 # index 1 to index 2 samma ko value lai 0 lay update gareko.
print(salary)

# New variable ma update gareko salary with old salary lai rakheko.
updated_salary = salary.copy() # updated_salary[1:4] = 0 # index 1 to index 3 ko salary lai 0 lay update gareko.
print("Updated_Salary:", updated_salary)

new_salary = salary.copy()
new_salary[:-5] = 0  # index -5 to 1st index ko salary lai 0 lay update garko.
print("New_Salary", new_salary)
```

```python
# Type 12: Slicing bata dherai value aakai patak update garne.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

new_salary = salary.copy()
new_salary[1:4] = [1000, 2000, 3000]
print("New Salary:", new_salary)

new_salary = salary.copy()
new_salary[-3:] = [1000, 2000, 3000]
print("New Salary:", new_salary)
```

```python
# Type 13: 2D Array ma Slicing.
   # Data Science ma data mainly rows and columns ma hunxan so 2D slicing pani important xa.

employees = np.array([[20000, 2], [30000, 3], [40000, 4], [10000,1], [50000, 5]])
print(employees) 

print(employees.flatten()) # output sabai aautai line ma aaunxa.
print(", ".join(map(str, employees.flatten())))  # sabai values lai comma le xuttyayera eutai line ma dekhaune.
print(" . ".join([f"{salary}, {experience}" for salary, experience in employees]))  # harek employee ko data comma ra dot le chuttyaune.

print(employees[0:2, :])

print(employees[:, 0]) # : = yeslay sabai salary nikalxa ra 0 index vanda aagadi samma ko years dekhaunxa.

print(employees[:, 1]) # sabai employees ko experience years matra nikalxa.

print(employees[1:3]) # index 1 ra index 2 ko data nikalxa.
```

                       2.3, Filtering ?
= Filtering vaneko condition ko aadhar ma array bata chahiyeko         values matra select garnu ho.

2.3.1, Filtering kina use garienxa?
- Rs. 30,000 bhanda badhi salary paune employees nikalna.
- 25 barsa bhanda mathi age bhayeka employees select garna.
- Kunai product ko price Rs. 5,000 bhanda kam xa ki xaina herna.

```python
# Operations of Filtering.
```

```python
# Operation 1: Greater than (>) Filtering. == Kunai value arko value bhanda thulo xa ki xaina check garna > use garinxa.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary) 

greater_than_30000_salary = salary[salary > 30000] # 30000 vanda badhi salary select garne.
print("Greater_Than_30000_Salary:", greater_than_30000_salary) 

greater_than_50000_salary = salary[salary > 50000]
print("Greater_Than_50000_Salary:",greater_than_50000_salary)
```

```python
# Operation 2: Smaller than (<) Filtering. == Kunai value arko value bhanda sano xa ki xaina check garna < use hunxa.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary) 

smaller_then_40000_salary = salary[salary<40000]
print("Smaller_Then_40000_Salary:", smaller_then_40000_salary)

smaller_then_30000_salary = salary[salary<30000]
print("Smaller_Then_30000_Salary:", smaller_then_30000_salary)
```

```python
# Operation 3: Greater than or equal (>=)

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

filtered_salary = salary[salary >= 40000]
print("Filtered_Salary:", filtered_salary)
```

```python
# Operation 4: Less than or equal (<=)

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

filtered_salary = salary[salary <= 30000]
print("Filtered_Salary:", filtered_salary)
```

```python
# Operation 5: Equal to (==)

salary = np.array([20000, 30000, 20000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

filtered_salary = salary[salary == 20000]
print("Filtered_Salary:", filtered_salary)

filtered_salary = salary[salary == 19000] # 19000 xaina tei vayara khali dekhaunxa.
print("Filtered_Salary:", filtered_salary)

filtered_salary = salary[salary == 20000]  # 20000 salary bhayeka values filter garne
print("Filtered_Salary:", ", ".join(map(str, filtered_salary)))  # filtered salary comma sahit dekhaune
```

```python
# Operation 6: Not equal to (!=)

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

Not_Equal_To_40000 = salary[salary != 40000]
print("Not_Equal_To_40000:", Not_Equal_To_40000)

Not_Equal_To_25000_and_18000 = salary[(salary != 25000) & (salary != 18000)]
print("Not_Equal_To_25000_and_18000:", Not_Equal_To_25000_and_18000)
```

```python
# Operation 7: True/False mask herne, Boolean Mask vaninxa.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

mask = salary > 30000 # 30000 vanda badi salary lai True dekhaunxa.
print(mask)

# aaba yo mask use garera values select garna sakienxa.
filtered_salary = salary[mask]
print("Filtered_Salary", filtered_salary)
```

```python
# Operation 8: Multiple conditions using (&)

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

new = salary[(salary<60000) & (salary>20000)]
print("New", new)

new = salary[(salary<=60000) & (salary>=20000)]
print("New:", new)
```

```python
# Operation 8: Multiple conditions using (|) == 2 condition ma kunai aak matra true vaya pani select garne.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

new = salary[(salary < 20000) | (salary> 50000)]
print("New:", new)
```

```python
# Operation 9: Not (~) use garne. == ~ le Boolean condition ko result ulto banaidinxa. True lai False ra False lai True banaunxa.

# Example: Rs. 30,000 bhanda badhi nabhayeka salary select garne.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

mask = salary > 30000  # 30000 bhanda badhi salary ko mask banaune

filtered_salary = salary[~mask]  # Mask ulto parera 30000 wa kam salary select garne
print(filtered_salary)  # Selected salary dekhaune
```

```python
# Operation 10: np.where() bata filtering == np.where() condition ko aadhar ma values ko index nikalna wa condition anusar different values return garna use hunxa.

# Example: Rs. 30,000 bhanda badhi salary kun-kun index ma xa?

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

index = np.where(salary > 30000)  # 30000 bhanda badhi salary ko index nikalne
print(index)  

# Aba condition anusar salary lai label garna pani sakinxa:

result = np.where(salary > 30000, "High", "Low")  # Salary 30000 bhanda badhi bhaye High natra Low rakhne
print(result)
```

```python
# Operation 11: Filtering bata original index ra salary dubai nikalne.

# Kahile kahi salary matra hoina, tyo salary array ko kun position ma xa vanne pani chahinxa.

salary = np.array([20000, 30000, 10000, 40000, 25000, 18000, 45000, 50000, 15000, 35000, 65000])
print(salary)

index = np.where(salary > 30000)[0]  # Condition pura garne index nikalne

filtered_salary = salary[index]  # Tyo index ko salary nikalne

print("Index:", index)  # Selected index dekhaune
print("Salary:", filtered_salary)  # Selected salary dekhaune
```

```python
# Operation 12: 2D array ma filtering.

employees = np.array([[20000, 1], [30000, 2], [10000, 1], [40000, 5], [50000, 6]])

filtered_employees = employees[employees[:, 0] > 30000]  # Pahilo column ko salary 30000 bhanda badhi bhayeka row select garne
print(filtered_employees)  # Selected employees ko complete row dekhaune
```

```python
# Different ways of iteration in NumPy.
```

```python
# 1. Normal python (for loop) bata Array iteration. = for loop le array bhitra ko harek element lai palaipalo access garxa.

salary = np.array([20000, 30000, 10000, 40000, 25000])

for amount in salary: # salary ko harek value lai amount variable ma rakhne.
    print(amount)
```

```python
# 2. Array ko elements ma calculation garne. = Iteration ko help le harek element ma calculation garna sakinxa.

# Suppose, hamilai harek employee ko salary ma Rs. 5,000 bonus add garera hernu xa.

salary = np.array([20000, 30000, 10000, 40000, 25000])
print(salary)

for amount in salary:
    print(amount + 5000)

print() # output bich ma gap dinxa.
print("Double Amount:")

for amount in salary:
    print(amount * 2)

print()
print("Salary divided by 2000:")

for amount in salary:
    print(amount / 2000)
```

```python
# 3. np.nditer() bata Array Iteration.

# np.nditer() NumPy ko built-in function ho, jasle array ko elements lai ek-ek garera iterate garna help garxa.

# Yo function 1D, 2D ra 3D jasta multidimensional arrays ma pani useful hunxa.

salary = np.array([20000, 30000, 10000, 40000, 25000])  # Salary ko 1D array banaune

for amount in np.nditer(salary):  # NumPy ko nditer bata harek element access garne
    print(amount)  # Current element dekhaune
```

```python
# 4. 2D Array ma Iteration

employees = np.array([[20000, 1], [30000, 2], [40000, 5]])  # Salary ra experience bhayeko 2D array banaune
print(employees)  # Pura 2D array dekhaune.

print()

# Normal for loop bata iterate garne.
for row in employees:
    print(row)

# Bujhne tarika: Normal for loop le 2D array ma harek choti euta row dinxa.
#- First iteration → [20000, 1]
#- Second iteration → [30000, 2]
#- Third iteration → [40000, 5]
```

```python
# 5. Array ko harek element alag alag iterate garne.
# salary ra experience dubai ko harek value alag-alag chahiyo vane np.nditer() use garna sakxau.

employees = np.array([[20000, 1], [30000, 2], [40000, 5]])  
print(employees)

print()

for value in np.nditer(employees): # # 2D array ko harek element ek-ek garera access garne
    print(value)
```

```python
# 6. 3D Array ma iteration.

# 3D array ma data multiple 2D arrays ko form ma hunxa. Yesto array image data, scientific data ra complex datasets ma use huna sakxa.
# np.nditer() le 3D array ko harek element lai pani iterate garna sakxa.

data = np.array([[[1, 2], [3,4], [5,6], [7,8]]]) # 3D array banaune.
for value in np.nditer(data): # 3D array ko harek elements access garne.
    print(value)

# Yaha 3D array bhayepani np.nditer() le sabai elements ek-ek garera access garyo.
```

```python
# 7. Iteration ko bela index ra value dubai nikalne.

# Kahile kahi Data Science ma value matra hoina, tyo value array ko kun index ma xa vanne pani chahinxa.
# Python ko (enumerate()) use garera index ra value dubai pauna sakinxa.

salary = np.array([20000, 30000, 10000, 40000, 25000])

for index, amount in enumerate(salary): # Harek element ko index ra value dubai nikalne.
    print("Index:", index, "Salary:",amount)
```

```python
# 8: Iteration bata total salary calculate garne.

# Iteration ko help le array ko values jodera total nikalna sakinxa. Tara NumPy ma total nikalna np.sum() jasto direct function pani xa.

import math

salary = np.array([20000, 30000, 10000, 40000, 25000])
total_salary = 0

for amount in salary: # harek salary access garne.
    total_salary = total_salary + amount # current salary lai total ma add garne.

print("Total_Salary:", total_salary)

print()

# large number multiply garna lai math function use garnu parxa, natra vane answer wrong huna salxa, output limit vanda thulo aayo vane overflow huna sakxa.
total_salary = math.prod(int(amount) for amount in salary)  # sabai salary ko exact product nikalne

print("Product_of_Salary:", total_salary)  # multiplication ko result dekhaune
```

```python
# 9: Iteration ko bela value update garne.

# At first, normal
salary = np.array([20000, 30000, 10000]) 
for amount in salary:
    amount = 0 # temporary variable ko value 0 banaune.
print(salary)

print()

# aaba index 1 ko salary update garna paryo.
salary[1] = 0
print(salary)
```

4.1, Copy and View?
=    NumPy ma array ko data duplicate garne wa existing data lai arko        tarikale access garne dui important concept hun:
- Copy = Original array ko data ko independent copy banaunxa. Copy ma     change garda original array change hudaina.
- View = Original data lai share garne arko array jasto access dinxa.     View bata data change garda original array pani change huna sakxa.    

. Data Science ma yo concepts data cleaning, preprocessing, feature       transformation ra original dataset safe rakhna important xan.

= Comparison of Copy and View
- Copy = Data independent.     / View = Usually shared data
- Copy = Copy change garda original no change. / View = View bata data change garda original change huna sakxa.
- Copy = Usually extra data memory chienxa.  / View = Usually data duplicate gardaina.
- Copy = .copy() / View = .view()    

```python
# 1. Copy() = Copy lay array ko data ko independent copy banaunxa.

salary = np.array([20000, 30000, 10000, 25000, 35000])

salary_copy = salary.copy() # Original array ko independent copy banaune.

print("Original Salary : ", salary)
print("Copy Salary : ", salary_copy)
```

```python
# 2. Copy ma change garda

salary = np.array([20000, 30000, 10000, 25000, 35000])
salary_copy = salary.copy()

salary_copy[2] = 50000 # copy salary ko index 2 lai 50000 lay replace gareko.

print("Original Salary : ", salary)
print("Copy Salary : ", salary_copy)
```

```python
# 3. Copy ma slicing garera update gareko.

salary = np.array([20000, 30000, 10000, 25000, 35000, 15000,  22000, 38000, 12000])
salary_copy = salary.copy()

salary_copy[1:4] = 5 # copy ko index 1 to index 3 samma ko salary lai 5 banaune.

print("Original Salary : ", salary)
print("Copy Salary : ", salary_copy)
```

```python
salary = np.array([20000, 30000, 10000, 25000, 35000, 15000,  22000, 38000, 12000])
salary_copy = salary.copy()

salary_copy[0:6:2] = 5
print("Copy : ", salary_copy)
```

```python
# 4. Copy ma calculation garne.

salary = np.array([20000, 30000, 10000, 25000, 35000, 15000,  22000, 38000, 12000])
print("Original Salary : ",salary)
salary_copy = salary.copy()

salary_copy = salary_copy + 1

print("New Salary : ", salary_copy)
```

```python
salary = np.array([20000, 30000, 10000, 25000, 35000, 15000,  22000, 38000, 12000])
print("Original Salary : ",salary)

salary_copy = salary.copy()

salary_copy[1:5] = salary_copy[1:5] + 1 # index 1 to index 3 ko salary ma matra change garna lai.

print("New Salary : ", salary_copy)
```

```python
# 4.2, View() 
# view() le original data share garne arko array banaunxa. Dui array ko shape wa data interpret garne tarika farak huna sakxa, tara normal NumPy array ko view le original data share garxa.
```

```python
# 1. Basic View

salary = np.array([20000, 30000, 10000, 25000, 35000, 15000,  22000, 38000, 12000])
print("Original Salary : ",salary)

salary_view = salary.view()

print("View:", salary_view)  # View array dekhaune
```

```python
# 2. View ma change garda

salary = np.array([20000, 30000, 40000]) 
salary_view = salary.view()  

salary_view[0] = 50000  # View ko first element change garne

print("Original:", salary)  # Original ma pani change bhayo ki herne
print("View:", salary_view)  # Updated view dekhaune
```

```python
# 3. View ma slicing garera update garne

salary = np.array([20000, 30000, 40000, 50000])  # Original salary array banaune
salary_view = salary[1:3]  # Original array ko slice view banaune

salary_view[0] = 0  # View ko first element zero banaune

print("Original:", salary)  # Original ma pani update bhayo ki herne
print("View:", salary_view)  # View ko updated values dekhaune
```

```python
# 4. View ma calculation ko result assign garne.

salary = np.array([20000, 30000, 40000])  # Original salary array banaune
salary_view = salary.view()  # Original data share garne view banaune

salary_view[:] = salary_view + 5000  # View ko sabai elements ma 5000 add garera assign garne

print("Original:", salary)  # Shared data ko karan original update bhayo ki herne
print("View:", salary_view)  # Updated view dekhaune
```

```python
# 5. Slicing paxi copy banaune

salary = np.array([20000, 30000, 10000, 25000, 35000, 15000,  22000, 38000, 12000])
print("Original Salary:", salary)

selected_salary = salary[1:5].copy()

selected_salary[3] = 1
print(selected_salary)
```

```python
# 6. Filtering bata copy ra view.

salary = np.array([20000, 30000, 10000, 25000, 35000, 15000,  22000, 38000, 12000])
print("Regular Salary:", salary)

high_salary = salary >= 30000 # high salary lai true ra low salary lai false dekhaunxa.
print(high_salary)

high_salary = salary[salary>=30000]
print(high_salary)
```

```python
salary = np.array([20000, 30000, 10000, 25000, 35000, 15000,  22000, 38000, 12000])
print("Regular Salary:", salary)

filtered_salary = salary[salary>=20000] # salary >= 20000 nikalxa.
print(filtered_salary)

filtered_salary[3] = 1 # filtered gareko salary ma vako index 3 ko salary lai 1 lay update garxa.
print(filtered_salary)

filtered_salary[1:4]  # filtered gareko salary ma vako index 1 to index 3 ko salary lai update garxa.
print(filtered_salary)
```

6.1, Array Transpose?
=    Array Transpose vaneko array ko rows lai columns ma ra columns lai       rows ma change garnu ho.

== Why it is use in Data Science?
-  Rows ra columns ko arrangement change garna.
-  Matrix calculations garna.
-  Linear algebra ra Machine Learning algorithms ma data ko shape           milaauna.
-  Numerical datasets ko dimensions manage garna.

```python
# 1: .T - Transpose garne sabai bhanda sajilo tarika.

# NumPy array ko .T attribute le rows ra columns ko position swap garxa. Yo 2D array ma frequently use hunxa.
```

```python
# Example 1: 

# Basic transpose.

numbers = np.array([[10, 20, 30], 
                    [40, 50, 60]
                   ])
transposed = numbers.T # rows ra columns swap garne.

print("Original:")
print(numbers)
print()
print("Transposed:")
print(transposed)
```

```python
# Salary and experience ko data transpose garne.

employees = np.array([[10000, 1], 
                      [20000, 2], 
                      [30000, 3], 
                      [40000, 4], 
                      [50000, 5], 
                      [60000, 6]
                     ])
transposed = employees.T
print("Original:")
print(employees)

print()

print("Transposed:")
print(transposed)
```

```python
# 2. np.transpose() function.

# .T ra np.transpose() dubai le array transpose garna sakxan. np.transpose() function use garda array lai argument ko rupma pass garinxa.
```

```python
# Example 2:

numbers = np.array([[10, 20], 
                    [30, 40], 
                    [50, 60]
                   ])
result_t = numbers.T  # .T use garera transpose garne
result_function = np.transpose(numbers)  # transpose() function use garera transpose garne

print("Using .T:")  # .T ko result ko heading dekhaune
print(result_t)  # .T ko result dekhaune

print()

print("Using np.transpose():")  # Function ko result ko heading dekhaune
print(result_function)  # Function ko result dekhaune

print(np.array_equal(result_t, result_function))  # Dubai result same xan ki check garne
```

```python
# 3: Square matrix transpose. row column ma column row ma.

# Square matrix vaneko rows ra columns ko number equal bhayeko matrix ho, jastai 2 × 2 wa 3 × 3. Square matrix transpose garda shape same rahanchha, tara elements ko position change hunxa.

matrix = np.array([[2,3], 
                   [4,5]
                  ])
transposed = matrix.T

print("Original:")
print(matrix)

print()

print("Transposed:")
print(transposed)

print("Shape", transposed.shape)
```

```python
# 4: 1D array transpose.

# 1D array ma euta matra dimension hunxa. Tesaile 1D array ma .T use garda pani shape change hudaina. 1D array lai column vector banauna reshape(-1, 1) use garna sakinxa.
```

```python
# Example 4: 1D array ma .T

numbers = np.array([10, 20, 30]) 

transposed = numbers.T  # 1D array ko transpose nikalne

print(transposed)  # Values dekhaune
print("Original shape:", numbers.shape)  # Original ko shape dekhaune
print("Transpose shape:", transposed.shape)  # Transpose ko shape dekhaune
```

```python
# 1D array lai column vector banaune.

numbers = np.array([10, 20, 30]) 
column_vector = numbers.reshape(-1,1)
print(column_vector)
```

```python
# 5: 3D array transpose.

# 3D array ma 3 ota axes/dimensions hunxan. Default .T le 3D array ma sabai axes ko order reverse garxa. Yo 2D transpose bhanda alik farak concept ho.
```

```python
# Example 5: 3D array ko shape.

data = np.array([[[1, 2, 3], [4, 5, 6]]])

transposed = data.T  # 3D array ko axes ko order reverse garne

print("Original shape:", data.shape)  # Original ko dimensions dekhaune
print("Transposed shape:", transposed.shape)  # Transpose paxi dimensions dekhaune
print(transposed)  # Transposed array dekhaune
```

7.1, Array Manipulation?
=    Array bhitra naya elements add garne, bichma elements insert garne,      ra unwanted elements delete garne tarika.

- Array Manipulation bhaneko NumPy array ko elements wa structure lai      change garnu ho. Jastai:

```python
# 1, np.append() - Array ma data add garne
```

```python
# Example 1: 1D array ko end ma aauta value add garne.

salary = np.array([20000, 30000, 40000])
new_salary = np.append(salary, 50000)

print("Original Salary:", salary)
print("New Salary:", new_salary)
```

```python
# Example 2: aakai choti multiple values add garne.
# Company ma 3 jana naya employee join vaya vane.

salary = np.array([20000, 30000, 40000])
new_salary = np.append(salary,[50000, 60000, 70000])

print("Original Salary:", salary)
print("New Salary:", new_salary)
```

Example 4: axis use garera 2D array ma row add garne.

2D array ma axis le kun direction ma data add garne bhanne janauchha.
- axis=0 → row ko direction ma add garne.
- axis=1 → column ko direction ma add garne.
- axis=None wa axis nadida → array flatten garera values add garne.

```python
# axis = 0 : row add garne.

employees = np.array([[20000, 1], 
                      [30000, 2]
                     ]) 
new_employees = [[40000, 4]]

updated_employees = np.append(employees, new_employees, axis=0)

print("Before Updated:")
print(employees)
print()
print("After Updated:")
print(updated_employees)

print()

print("Dimension:", employees.ndim)
print("Dimension:", updated_employees.ndim)
```

```python
# axis = 1 : Column add garne. like bonus add garnu paryo vane.

# 2 rows bhayeko original array ma pani 2 rows nai bhayeko bonus column add gareko xa.

employees = np.array([[1000, 1], 
                      [2000, 2]
                     ])
bonus = np.array([[2000], [2000]])
updated_employees = np.append(employees, bonus, axis=1)

print("Before Updated:")
print(employees)
print()
print("After Updated:")
print(updated_employees)
```

```python
# Example 1: Single element delete garne.

salary = np.array([20000, 30000, 40000, 5000])
new_salary = np.delete(salary, 2) # index 2 ma vako salary delete garxa.`
print("Before Salaey:", salary)
print("After Salary:", new_salary)

print ()

# Example 2: Multiple element delete garne.

numbers = np.array([20, 30, 40, 50, 60, 70, 80, 90])
new_numbers = np.delete(numbers, [1, 3, 5]) # index 1,3,5 ma vako numbers haru delete garxa.
print("Original Numbers:", numbers)
print("New Numbers:", new_numbers)
```

```python
# np.concatenate()

# yeslay 2 or more arrays lai existing axis (new line/axis na thapne) ko direction ma jodxa garxa.
```

```python
# Example 1: np.concatenate() with 1D array.

salary_1 = np.array([20000, 30000, 40000])
salary_2 = np.array([25000, 35000, 45000])

combined_salary = np.concatenate((salary_1, salary_2))
print(combined_salary)
```

```python
# Example 2: np.concatenate() with 2D array

# Row add garne.

# axis=0 by default hunxa. code ma lekhi rakhnu pardaina.

arr1 = np.array([[10, 20], 
                 [30, 40]
                ])
arr2 = np.array([[50, 60],
                [70, 80]
                ])

combined_arr = np.concatenate((arr1, arr2))

print("arr1:")
print(arr1)
print("arr2:")
print(arr2)
print()
print("Combined_arr:")
print(combined_arr)
```

```python
# Example 3: np.concatenate() with 2D array

# axis=1 - column add garne. code ma lekhnu parxa.

arr1 = np.array([[10, 20],
                [30, 40],
                 [50, 60]
                ])
arr2 = np.array([[15, 25],
                [35, 45],
                [55, 65]
                ])

combined_arr = np.concatenate((arr1, arr2), axis=1) # axis=1 lay column ma add garxa.

print("arr1:")
print(arr1)
print("arr2:")
print(arr2)
print()
print("Combined Arr:")
print(combined_arr)
```

```python
# Example 1: np.where() use garera value khojne.

numbers = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])
result = np.where(numbers == 40) # 40 kun index ma xa vanera dekhaunxa.
print(result)

print()

numbers = np.array([10, 40, 30, 40, 50, 60, 40, 80, 90])
result = np.where(numbers == 40) # 40 kun index ma xa vanera dekhaunxa.
print(result)
```

```python
# Example 2: Condition use garne

# >
numbers = np.array([10, 40, 30, 40, 50, 60, 40, 80, 90])
result = np.where(numbers > 35)
print(">35:",result)

print()

# <
numbers = np.array([35, 10, 40, 30, 40, 50, 60, 40, 80, 90])
result = np.where(numbers < 35)
print("<35:",result)

print()

# >=
numbers = np.array([35, 10, 40, 30, 40, 50, 60, 40, 80, 90])
result = np.where(numbers >= 35)
print(">=:",result)

print()

# <=
numbers = np.array([35, 10, 40, 30, 40, 50, 60, 40, 80, 90])
result = np.where(numbers <= 35)
print("<=:",result)
```

```python
# Example 1: Ascending order ma sort garne.

numbers = np.array([10, 8, 30, 58, 4, 57, 72, 44, 72, 39, 83, 47])
result = np.sort(numbers)
print("In Ascending:", result)
```

```python
# Example 2: Descending order ma sort garne.

# np.sort(numbers)[::-1] = first ascending ma sort garxa ani teslai reverse garxa.

numbers = np.array([10, 8, 30, 58, 4, 57, 72, 44, 72, 39, 83, 47])
result = np.sort(numbers)[::-1] # ::-1 slicing method le array ko sabai element lai ulto order ma rakhxa.
print("In Descending:", result)
```

```python
# Example 3: Condition use garera sort garne.

# num >= 30, num <= 50 , in ascending
numbers = np.array([10, 8, 30, 58, 4, 57, 72, 44, 72, 39, 83, 47])
result = np.sort(numbers[(numbers>=30) & (numbers<=50)])
print("In Ascending:",result)

# num >= 30, num <= 50 , in descending
numbers = np.array([10, 8, 30, 58, 4, 57, 72, 44, 72, 39, 83, 47])
result = np.sort(numbers[(numbers>=30) & (numbers<=50)])[::-1]
print("In Descending:",result)
```

Solving Linear Equation.

* Linear equations bhaneko ke ho? 
- Linear equations bhaneko unknown variables bhayeka equations hun, jastai:
- eg: (2x + y = 8) / (x + y = 5)
- Yo equations lai matrix form lekhna sakienxa. (A, X, B) ko form ma.
- ( np.linalg.solve() ) use garera unknown values nikalna sakxau.

```python
# Example 1: 

# Equations are: (2x + y = 8) & (x + y = 5)

a = np.array([[2, 1], [1, 1]]) # eqn 1 ma (x=2, y=1) xa and eqn 2 ma (x=1, y=1) xa.
b = np.array([8, 5]) # equal to paxi ko value (8 and 5).
result = np.linalg.solve(a, b)
print("Value of x and y:", result)

# Check garna lai aayako answer lai equation ma halne.
```

```python
# Example 2: 

# Equations are: (17x + 19y = 55.5) & (11x + y = 29.8)

a = np.array([[17, 19], [11, 1]])
b = np.array([55.5, 29.8])
result = np.linalg.solve(a,b)
print(result)
```

1. Random Number Generation vaneko ke ho?

-  Random Number Generation vaneko computer ko help le random jasto dekhinay numbers generate garnu ho.

2. Real life example:
-  Suppose 100 jana students ko data xa. among them, random 10 jana students select garna paryo. Random generation ko concept use garera random 10 jana students select garna sakinxa.

3. Data Science ma kaha use hunxa?
-  Data Testing: Testing ko lagi sample data banauna.
-  Machine Learning: Model testing ra simulation ko lagi random data generate garna.
-  n ko lagi random data generate garna.
-  Statistics: Random samples select garna.
-  Simulation: Real-world situations ko model banauna.
-  Data Analysis: Different scenarios ko result analyze garna.

Important:
1. np.random bhitra different functions xan. like,
-  np.random.randint()	Random integer generate garxa.
-  { np.random.rand() } or { np.random.random() }	0 dekhi 1 bhanda sano random decimal generate garxa. eg: 0.25, 0.78, 0.43.
-  np.random.uniform()	Given range bhitra random decimal generate garxa.
-  np.random.choice()	Available values madhye random value select garxa.
-  np.random.shuffle()	Array ko order change garxa.
-  np.random.permutation()	Array ko shuffled copy/permutation dinxa.
-  np.random.seed()	Random sequence repeatable banauna help garxa.

```python
# Example 1: np.random.randint() - Random integer generate garxa.

# 5 random no find garne. (or) 1D array generate garne.
numbers = np.random.randint(1, 100, size=5)  # 1 dekhi 99 samma ka 5 random integer banaune.
print("Random numbers:",numbers) # harek time print garda different random number aaunxa.
print("Shape:", numbers.shape)
print("Dimension:", numbers.ndim)
print("Random no mul by 2:",numbers * 2)

print()

# Single random no find garne.
number = np.random.randint(1, 20) # 1 to 20 samma aauta random number dinxa.
print("Random no:", number)
print("Random no mul by 2:", number * 2)

print()

# 4 random no find garne. (or) 1D array generate garne.
number = np.random.randint(40, 50, size=4) # 40 to 50 ko bicch bata 4 ota random number dinxa.
print("4 Random Numbers:", number)

print()

# 2D random array generate garne.
numbers = np.random.randint(1, 51, size=(3,4)) # 1 to 50 samma ko random number vako 3 row and 4 column banaune.
numbers = np.random.randint(1, 51, size=(3,4)) * 2
print("2D Random array:")
print(numbers)
print("2D Random array mul by 2:")
print(numbers)
```

```python
# Example 2: np.random.rand(): - 0 dekhi 1 bhanda sano random decimal generate garxa.

number = np.random.rand() # single random number gererate garxa.
print("Single Random No:", numbers)
print("Afert multiplication:", numbers * 0)

print()

# 5 random no (or) 1D array:
numbers = np.random.random(5)
print("Five Random No:", numbers)

print()

# random 2D array:
numbers = np.random.random((3,4))
print("Random 2D array:")
print(numbers)
```

```python
# Example 3: np.random.choice() - diyeko values haru madhye random tarikale euta wa multiple values select garxa.

# single random no/value/person find garne.
students = np.array(["Ram", "Shyam", "Hari", "Sita", "Gita"])
selected = np.random.choice(students)  # students madhye euta random student select garne
print("Single Random Student:",selected)

print()

# 3 random no/value/person find garne.
students = np.array(["Ram", "Shyam", "Hari", "Sita", "Gita"])
selected = np.random.choice(students, size=3)  # 3 random select garne.
print("3 Random Student:",selected)

print()

# 3 random no/value/person find garne.
students_id = np.array([10,30,42,23,24,65])
selected = np.random.choice(students_id, size=3)  # 3 random select garne.

print("3 Random Student_id:",selected)
```

```python
# Duplicate values aauna nadina.
# repeat navayeka values chahiyeko xa bhane replace=False use garne.

numbers = np.array([10, 20, 30, 40, 50])  
selected = np.random.choice(numbers, size=3, replace=False)  # repeat nagari 3 ota values select garne
print(selected)

# replace=True → eutai value feri select huna sakxa.
# replace=False → select vako value feri select hudaina.
```
